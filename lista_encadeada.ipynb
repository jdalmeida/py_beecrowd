# Lista encadeada (singly linked list)

Implementação de uma lista encadeada simples em Python: nós, inserção, remoção e percorrimento.


```python
# @author João Gabriel de Almeida
from __future__ import annotations

from dataclasses import dataclass
from typing import Any, Iterator, Optional


@dataclass
class No:
    valor: Any
    proximo: Optional["No"] = None

class ListaEncadeada:
    def __init__(self) -> None:
        self._head: Optional[No] = None
        self._size: int = 0

    def __len__(self) -> int:
        return self._size

    def is_empty(self) -> bool:
        return self._head is None

    def adicionar_inicio(self, valor: Any) -> None:
        self._head = No(valor, self._head)
        self._size += 1

    def adicionar_fim(self, valor: Any) -> None:
        novo = No(valor)
        if self._head is None:
            self._head = novo
        else:
            atual = self._head
            while atual.proximo is not None:
                atual = atual.proximo
            atual.proximo = novo
        self._size += 1

    def inserir_em(self, indice: int, valor: Any) -> None:
        if indice < 0 or indice > self._size:
            raise IndexError("índice fora do intervalo")
        if indice == 0:
            self.adicionar_inicio(valor)
            return
        anterior = self._no_em(indice - 1)
        anterior.proximo = No(valor, anterior.proximo)
        self._size += 1

    def remover_inicio(self) -> Any:
        if self._head is None:
            raise IndexError("lista vazia")
        valor = self._head.valor
        self._head = self._head.proximo
        self._size -= 1
        return valor

    def remover_fim(self) -> Any:
        if self._head is None:
            raise IndexError("lista vazia")
        if self._head.proximo is None:
            return self.remover_inicio()
        anterior = self._head
        while anterior.proximo.proximo is not None:
            anterior = anterior.proximo
        valor = anterior.proximo.valor
        anterior.proximo = None
        self._size -= 1
        return valor

    def remover_em(self, indice: int) -> Any:
        if indice < 0 or indice >= self._size:
            raise IndexError("índice fora do intervalo")
        if indice == 0:
            return self.remover_inicio()
        anterior = self._no_em(indice - 1)
        removido = anterior.proximo
        assert removido is not None
        valor = removido.valor
        anterior.proximo = removido.proximo
        self._size -= 1
        return valor

    def obter(self, indice: int) -> Any:
        return self._no_em(indice).valor

    def contem(self, valor: Any) -> bool:
        atual = self._head
        while atual is not None:
            if atual.valor == valor:
                return True
            atual = atual.proximo
        return False

    def __iter__(self) -> Iterator[Any]:
        atual = self._head
        while atual is not None:
            yield atual.valor
            atual = atual.proximo

    def __repr__(self) -> str:
        partes = [repr(v) for v in self]
        return "ListaEncadeada([" + ", ".join(partes) + "])"

    def _no_em(self, indice: int) -> No:
        if indice < 0 or indice >= self._size:
            raise IndexError("índice fora do intervalo")
        atual = self._head
        for _ in range(indice):
            assert atual is not None
            atual = atual.proximo
        assert atual is not None
        return atual
```

## Exemplos

```python
lst = ListaEncadeada()
lst.adicionar_fim(1)
lst.adicionar_fim(2)
lst.adicionar_inicio(0)
print(lst, "tamanho:", len(lst))

lst.inserir_em(2, 99)
print("após inserir 99 no índice 2:", lst)

print("obter(1):", lst.obter(1))
print("contém 99?", lst.contem(99))

print("remover_fim:", lst.remover_fim(), "→", lst)
print("remover_inicio:", lst.remover_inicio(), "→", lst)
print("list(...):", list(lst))
```
